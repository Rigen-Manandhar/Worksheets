#3 Image Compression and Decompression using PCA.
In this exercise, build a PCA from scratch using explained variance method for image compression task.

##1. Load and Prepare Data:

    • Fetch an image of you choice.{If colour convert to grayscale}

    • Center the dataset - Standaridze the Data.

    • Calculate the covaraince matrix of the Standaridze data.

```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# load the grayscale image
img = Image.open("lena_gray.gif")

# convert image to numpy array
img_array = np.array(img, dtype=float)

# display the image
plt.imshow(img_array, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()

# standardize the data
mean = np.mean(img_array, axis=0)
std = np.std(img_array, axis=0)

# avoid division by zero
std[std == 0] = 1

standardized_data = (img_array - mean) / std

# calculate covariance matrix
cov_matrix = np.cov(standardized_data, rowvar=False)

print("Image shape:", img_array.shape)
print("Standardized data shape:", standardized_data.shape)
print("Covariance matrix shape:", cov_matrix.shape)
```

#2. Eigen Decomposition and Identifying Principal Components:

    • Compute Eigen Values and Eigen Vectors.

    • Sort the eigenvalues in descending order and choose the top k eigenvectors corresponding to the highest eigenvalues.

    • Identify the Principal Components with the help of cumulative Sum plot.

```python
import numpy as np
import matplotlib.pyplot as plt

# compute eigenvalues and eigenvectors
eigen_values, eigen_vectors = np.linalg.eigh(cov_matrix)

# sort eigenvalues in descending order
sorted_index = np.argsort(eigen_values)[::-1]
sorted_eigen_values = eigen_values[sorted_index]
sorted_eigen_vectors = eigen_vectors[:, sorted_index]

# explained variance
explained_variance = sorted_eigen_values / np.sum(sorted_eigen_values)

# cumulative sum of explained variance
cumulative_variance = np.cumsum(explained_variance)

# plot cumulative sum
plt.plot(cumulative_variance)
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Sum Plot")
plt.grid()
plt.show()

# choose top k eigenvectors
k = np.argmax(cumulative_variance >= 0.95) + 1
top_k_eigenvectors = sorted_eigen_vectors[:, :k]

print("Number of principal components selected:", k)
print("Shape of top k eigenvectors:", top_k_eigenvectors.shape)
```

##3. Reconstruction and Experiment:

    • Reconstruction: Transform the original data by multiplying it with the selected eigenvec- tors(PCs) to obtain a lower-dimensional representation.

    • Experiments: Pick Four different combination of principal components with various ex-plained variance value and compare the result.
    
    • Display the Results and Evaluate.

```python
import numpy as np
import matplotlib.pyplot as plt

# choose four explained variance values
variance_values = [0.60, 0.75, 0.90, 0.95]

plt.figure(figsize=(10, 10))

for i, var in enumerate(variance_values):
    # find number of principal components needed
    k = np.argmax(cumulative_variance >= var) + 1

    # select top k eigenvectors
    selected_vectors = sorted_eigen_vectors[:, :k]

    # transform to lower-dimensional representation
    reduced_data = np.dot(standardized_data, selected_vectors)

    # reconstruct the image
    reconstructed_data = np.dot(reduced_data, selected_vectors.T)

    # convert back to original scale
    reconstructed_data = reconstructed_data * std + mean

    # display the reconstructed image
    plt.subplot(2, 2, i+1)
    plt.imshow(reconstructed_data, cmap="gray")
    plt.title(f"Variance = {int(var*100)}%, PCs = {k}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
